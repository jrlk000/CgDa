### Versuchsaufbau
---
In dieser Beispielaufgabe <mark>bestimmen Sie die Erdbeschleunigung $g$</mark> mithilfe einer Feder.
Die Bestimmung erfolgt dabei in zwei Schritten:
- In einem ersten Schritt wird dazu zunächst aus einer Messung die <mark>Federkonstante $D$</mark> der Feder bestimmt:
Eine bekannte Masse $M$ wird an der Feder angebracht und in Schwingung versetzt.
Aus der Periodendauer der Schwingung lässt sich $D$ ermitteln.
- Mit dieser Information kann in einem zweiten Schritt aus einer weiteren Messreihe $g$ bestimmt werden: 
Es wird die <mark>Auslenkung $s$ </mark> derselben Feder unter Last sehr genau bekannter Massen $m_i$ gemessen.
Durch die Anpassung des bekannten Zusammenhangs zwischen Auslenkung und wirkender Kraft an diese Messdaten wird $g$ bestimmt.

Die Daten für den ersten Schritt wurden durch Auslesen des Beschleunigungssensors eines Mobiltelefons mit der App [`phyphox`](https://phyphox.org) gewonnen. 
Das Mobiltelefon mit der Masse $M=141,74\,\mathrm{g}$ wurde dazu an der Feder mit der Masse $M_\mathrm{F}=15,40\,\mathrm{g}$ befestigt und in vertikale Schwingungen versetzt.
Für die zweite Messreihe wurden verschiedene Gewichte an dieselbe Feder gehängt und die Auslenkung mithilfe eines Lineals und "gutem Augenmaß" bestimmt.

**Hinweis:**<br>
Das gewichtete Mittel der Messwerte $x_i$ mit Gewichten $w_i=1/\sigma_i^2$, wobei $\sigma_i$ der Unsicherheit auf $x_i$ entspricht, berechnet sich aus
$$
\bar{x}=\frac{\sum_i w_i x_i}{\sum_i w_i}.
$$
Die Varianz $\sigma^2_{\bar{x}}$ auf den gewichteten Mittelwert $\bar{x}$ ergibt sich durch lineare Fehlerfortpflanzung als:
$$
\sigma^2_{\bar{x}}=\frac{\sum_i w_i^2 \sigma_i^2}{\left(\sum_i w_i\right)^2}=\frac{1}{\sum_i \frac{1}{\sigma_i^2}}.
$$

```python
# ----IMPORTS----
import numpy as np
import matplotlib.pyplot as plt
import PhyPraKit as ppk


# ----Aufbereitung der Messdaten----

data = np.genfromtxt('HandyPendel.csv', delimiter=',', names=True, dtype=None)

t, x, y, z, abs = data.dtype.names

times = data[t]
x_acc = data[x]
y_acc = data[y]
z_acc = data[z]
abs_acc = data[abs]

accelerations = (x_acc, y_acc, z_acc)
a = set(x_acc) | set(y_acc) | set(z_acc)
y_max = max(a)
y_min = min(a)

"""a = max(y_acc)
b = min(y_acc)

c = max(z_acc)
d = min(z_acc)

e = max(x_acc)
f = min(x_acc)

print(a, b, c, d, e, f)"""

"""
Wir erkennen aus dem Datensatz, dass die Beschleunigungen in die x-Richtung im Bereich
1.595077276 -1.692552924 liegen und die Beschleunigungne in der z-Achse im Bereich von  0.7445577383 -1.626938105 liegen, was uns schließen lässt, dass die Schwingung primär in der y-Ebene passiert.

Die Beschleunigung in der y-Ebene beläuft sich auf den Bereich von 7.370850563 -7.40380621.
"""

# ----Graphische Darstellung der Beschleunigung in die Raumrichtungen.
fig, axes = plt.subplots(3, 1)

#Anpassunge an der Figure.
fig.suptitle("Beschleunigung in die Raumrichtingen.", y=1 ,fontsize=18)
fig.set_size_inches(10, 10)

#Erstellen der einzelnen Axes Objekte
for ax, acc, name in zip(axes, accelerations, data.dtype.names[1:4]):

    ax.plot(times, acc, color='black', linewidth=4, label=name)

    #Nomiere die Ausschläge bassierend auf dem größten Ausschlag
    ax.set_ylim(y_min, y_max)

    # Anapssunegn am ax Objekt
    ax.set_aspect('auto')
    ax.grid(alpha=0.4) #Slightly lighter grid
    ax.set_xlabel("Time in sek.", fontsize=12)
    ax.set_ylabel("Beschleunigung in [m/s²]", fontsize=12)
    #ax.legend(loc="upper left", fontsize=12)


plt.tight_layout()
plt.show()


#----Finden der Maxima aus den Datensätzen----

#Relevant ist hier lediglich die Beschleunigungs Peaks in dien tatsächliche Schwingungsrichtug y.

# Um den Peakfinder von PhyPraKit richtig verwenden zu könne, muss angegeben werden wie viel Massepunkte ca. zwischen den Peaks sind. Hier: 25 bis 35

breiten_bereich = 40

peak_indizes_high = ppk.convolutionPeakfinder(y_acc, breiten_bereich)
peak_indizes_low = ppk.convolutionPeakfinder(-y_acc, breiten_bereich)

fig, ax = plt.subplots()
fig.set_size_inches(10, 10)
ax.plot(times, y_acc, color='black', label='y_Acc.')
ax.set_xlabel("Times in sek.")
ax.set_ylabel("Acc in m/s²")
ax.grid()

ax.plot(times[peak_indizes_high], y_acc[peak_indizes_high], "ro", markersize=8, label='Peaks')
ax.plot(times[peak_indizes_low], y_acc[peak_indizes_low], "ro", markersize=8, label='Peaks')

plt.tight_layout()
plt.show()

#Ploten des Histograms für die Zeitdifferenzen der Hochpunkte und der Tiefpunkte der Beschleunigungsmesswerte.

#Highs
fig, (ax1, ax2) = plt.subplots(2, 1)
fig.set_size_inches(10, 20)
zeitdifferenzen_high = np.roll(times[peak_indizes_high], -1)[:-1] - times[peak_indizes_high][:-1]
ax1.hist(zeitdifferenzen_high, bins='auto', color='blue', edgecolor='black')

ax1.set_title("Verteilung der Schwingungsdauern")
ax1.set_xlabel("Zeitdifferenz zwischen Maxima [s]")
ax1.set_ylabel("Häufigkeit")

#Lows
zeitdifferenzen_low = np.roll(times[peak_indizes_low], -1)[:-1] - times[peak_indizes_low][:-1]
ax2.hist(zeitdifferenzen_low, bins='auto', color='blue', edgecolor='black')

ax2.set_title("Verteilung der Schwingungsdauern")
ax2.set_xlabel("Zeitdifferenz zwischen Minima [s]")
ax2.set_ylabel("Häufigkeit")

plt.show()

"""
gew_mean = lambda w, x: np.sum(w*x/w)
"""
periodendauer_high_mean = np.mean(zeitdifferenzen_high)
var_high = np.var(zeitdifferenzen_high)

periodendauer_low_mean = np.mean(zeitdifferenzen_low)
var_low = np.var(zeitdifferenzen_low)

"""gesamt = np.concatenate((zeitdifferenzen_high, zeitdifferenzen_low))"""

#----gemittelten Wertes der Periodendauer---
# (aus Hochpunkten und aus Tiefpunkten.)# Berechne zuerst die normalen Mittelwerte
mean_high = np.mean(zeitdifferenzen_high)
mean_low = np.mean(zeitdifferenzen_low)

w_high = 1 / var_high
w_low = 1 / var_low

# Die saubere Formel für den gewichteten Mittelwert:
gew_mean_periodendauer = (w_high * mean_high + w_low * mean_low) / (w_high + w_low)

#----Varianz des gewichttetn Mittelwertes----
varianz_gewichteten_mittelwertes_periodendauer = var_high*var_low/(var_high+var_low)
unsicherheit_periodendauer = np.sqrt(varianz_gewichteten_mittelwertes_periodendauer)

#Werte für die Frequenz (gemittelter MIttelwert, Stichprobenvarianz)
gemittelte_frequenz = 1/gew_mean_periodendauer
stichprobenvariance_frequence = 1/(gew_mean_periodendauer**2) * varianz_gewichteten_mittelwertes_periodendauer
unsicherheit_frequenz = np.sqrt(stichprobenvariance_frequence)


#Bestimmung der Federkonstnet

federkonstante = lambda x, y: (x*2*np.pi)**2*y

m_eff = (141.74 + 15.40/3)/1000 # [kg]
#Unsicherheit ergibt sich zu delta m_eff  = delta m + 1/3 * delta m_feder

gemittelte_Federkonstante = federkonstante(gemittelte_frequenz, m_eff)

#[Beachte]: Unsicherheit des Gerätes +|-0.1g und Unsicherheit der Frequenz ergibt sich aus Stichprobenstandardabweochung

#Unsicherheit ergibt sich zu delta m_eff  = delta m + 1/3 * delta m_feder
unsicherheit_m_eff_kg = ((1e-1+1e-1*1/3)/1000)

# Fehlerfortpflanzung für D
teil_f = (8 * np.pi**2 * gemittelte_frequenz * m_eff) * unsicherheit_frequenz
teil_m = (4 * np.pi**2 * gemittelte_frequenz**2) * unsicherheit_m_eff_kg

unsicherheit_Federkonstante = teil_f + teil_m

print(f"Die Federkonstante er gibt sich zu: {gemittelte_Federkonstante} +/- {unsicherheit_Federkonstante}")










```

Durchführen der Parameterschätzung für den Wert der Erdbeschleunigung. Unter Verwendung von kafe2.

```python
def auslenkung(m:float, D:float=D, g: float=9.81)->float:
    """

    """
    return m*g/D

fit = XYFit(xy_data=[m, s], model_function=auslenkung)

#Fehler der Größen m, s bestimmen.
fit.add_error(axis='x', err_val=unsicherheit_massen)
fit.add_error(axis='y', err_val=unsicherheit_auslenkung)

#Berechnete Wert, Unsicherheit für die Federkonsttante als Referenz eingeben

fit.add_parameter_constraint(name='D', value=D, uncertainty=unsicherheit_D)

fit.assign_parameter_latex_names(D='D', m='M', g='g')

fit.assign_model_function_latex_expression(
    r"\frac{{mg}}{{D}}"
)

# Fit ausführen
fit.do_fit()
# Optional: Ausführlicher Bericht über den Fit
fit.report()
# Darstellung des Fits im kafe20Stil
plot = Plot(fit)
plot.x_label = "$m [kg]$"
plot.y_label = "$s [m]$"
plot.plot()
plot.show()
```
