**Digitale Signalverarbeitung im Praktikum**<br>

Messgeräte im physikalischen Praktikum wandeln häufig die Messgröße in analoge elektrische Signal um, z. B. in elektrische Spannungen. Diese <mark>analogen Signale sind **kontinuierlich**</mark>, sowohl als Funktion der Zeit als auch als Funktion des Signalwerts. Um ein analoges Signal in ein digitales Signal umzuwandeln, muss es mithilfe eines <mark>**Analog-Digital-Wandlers**</mark> (engl.:analog-to-digital converter, ADC) sowohl in der Zeit als auch im Signalwert in diskrete Werte umgewandelt werden. Solche ADCs sind häufig direkt in den Messgeräten im physikalischen Praktikum verbaut. Sie liefern die Messdaten als **Zeitreihen**, mit einem oder mehreren Messwerten zu festen Zeitpunkten.

Einfach strukturierte digitale Daten wie solche Messdaten lassen sich in Form von Tabellen aus Gleitkommazahlen abspeichern. Ein sehr gängiges Format für solche Daten ist das <mark>**CSV-Format**</mark> (engl.: comma-separated values), das Sie bereits im Kurs "Programmieren und Algorithmen" kennengelernt haben. Eine solche CSV-Datei mit einem digitalisierten Messsignal soll in dieser Aufgabe eingelesen und verarbeitet werden. Die ersten fünf Zeilen dieser Datei (`signal_raw.csv`) sehen wie folgt aus:
```
time (ms),amplitude(V)
0,-0.0591771826
0.05564830273,0.1081948399
0.1112966055,0.2375716725
0.1669449082,0.03774585915
```
Wie der Name CSV suggeriert, sind die Spalten der Tabelle durch Kommata getrennt. Die erste Zeile ist die Kopfzeile, sie enthält als **Metadaten** die Größen, die in den Spalten dargestellt werden und deren Einheiten. Alle weiteren Zeilen enthalten die Messdaten.

Mit dem Begriff <mark>**Filterung**</mark> wird in der Signalverarbeitung das **absichtliche Verändern** von Daten unter Ausnutzung von **Vorwissen** über Signal und Rauschen bezeichnet. Bei der <mark>**Glättung**</mark> von Signalen wird dabei ausgenutzt, dass sich das Signal nur langsam innerhalb des Datensatzes ändert, während sich das Rauschen schneller ändert. Ein einfaches Filter zur Glättung verwendet den <mark>**gleitenden Mittelwert**</mark> der Daten ("Mittelwertfilter"). Ein digitales Signal aus $N$ Datenpunkten $y_i$ ($i=0,\dots,N$) besteht, wird dabei wie folgt gefiltert:
$$
m_j = \frac{1}{2k+1} \sum_{i = -k}^{k} y_{i+j}.
$$
Das neue Signal $m_j$ besteht also aus dem arithmetischen Mittel über $w=2k+1$ Werte – den Wert $y_i$ und jeweils $k$ Werten links und rechts davon. Als Glättungsparameter wird dann der Parameter $k$ oder $w$ bezeichnet, er muss passend zu Signal und Rauschen gewählt werden.

1. Zum einfachen visuellen Vergleich verschiedener Glättungsparameter bietet es sich an, das Glätten und Darstellen des Ergebnisses und der Differenz in einer Funktion zu definieren.

In Kapitel 4 der Vorlesung wird das Python-Paket [`PhyPraKit`](https://etpwww.etp.kit.edu/~quast/PhyPraKit/htmldoc/) vorgestellt, das auf den Jupyter-Servern des KIT vorinstalliert ist und mit `import PhyPraKit as ppk` eingebunden werden kann. Es enthält ebenfalls eine Implementation eines Mittelwertfilters: [`PhyPraKit.meanFilter(a, width)`](https://github.com/GuenterQuast/PhyPraKit/blob/94779a6fdf6958fac34d8af425644129c1cbc6ff/PhyPraKit/phyTools.py#L872). Hier ist `a` das Array des jeweiligen Signals und `width` ist $w=2k+1$ und bestimmt den Glättungsparameter. 


````python
# Implementation des Mittelwertfilters
def moving_average(signal, width):
    """compute moving average of dataset averaging over 2k+1 data points"""
    filter = np.ones(width) / width
    return np.convolve( signal, filter, mode='same')

# verschiedene Glättungsparameter
widths = [k for k in (0, 4, 12, 100)]

#geglättete Signale
geglätete_signale = ([amplitudes] +
                     [moving_average(amplitudes, w) for w in widths[1:]])

differenz = [np.array(amplitudes, dtype=float) - g for g in geglätete_signale]

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))

for ax, ampl, w, d in zip(axs.flatten(), geglätete_signale, widths, differenz):
    ax.plot(times, ampl, color='blue', label='geglätetes Signal')
    ax.plot(times, d, color="red", label="Diff. zu Messung")

    ax.set_title(f'Glättungsparameter w: {w}')
    ax.set_xlabel("Zeit")
    ax.set_ylabel("Amplitude")
    ax.grid(True)
    ax.legend(loc='upper left')

plt.tight_layout()
plt.show()

"""
was wir erkennen aus den verschiedenen Werten für w:
Kleine Werte für den Glättungsparameter sorgen dafür, dass dsa Signal nach wie vor nicht einer idealean sinusförmigen Schwingung folgt, sonder kleine Einbrüche oder Übersteigerungen des optimalen wertes erkenntlich sind.

Große Werte für den Glättungsparameter hingegen sorgen dafür, dass das Signal weggeglätet wird also die eigentlichen Ausschläge des Signals duch das bilden des Mitelwertes über eine großen Bereich nach unten gezogen werden.
"""
````

Sie betrachten in dieser Aufgabe die gemessenen Daten einer Resonanzkurve, welche durch folgendes Modell beschrieben wird:
$$
A(\eta; D)=\frac{1}{\sqrt{(1-\eta^2)^2 + (2\eta D)^2}} \,.
$$
Gemessen wurde die Amplitude $A$ einer erzwungenen Schwingung in Abhängigkeit der normierten Frequenz $\eta\equiv\omega/\omega_0$.
Die Messdaten und deren Unsicherheiten sowie Resonanzkurven für verschiedene Werte von $D$ werden in den folgenden Zellen für Sie dargestellt.

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Resonanzfunktion
def resonance(eta, D):
    return 1/np.sqrt((1 - eta*eta)**2 + (2*eta*D)**2)

# Gemessene Datenpunkte
etam = np.array([0.10, 0.30, 0.50, 0.70, 0.90, 1.10, 1.30, 1.50, 1.70, 1.90, 2.10])
Am = np.array([0.97, 0.92, 1.12, 1.09, 1.36, 1.17, 0.76, 0.54, 0.47, 0.41, 0.36])

# Unsicherheit auf die gemessene Amplitude
Aerr = 0.1
````

````python
# Darstellung an Funktionenschar der Resonanzkurve
etamin, etamax = 0, 3
npoints = 1000
eta = np.linspace(etamin, etamax, npoints, endpoint=True)

# Darstellung der verschiedenen Resonanzen
for D in (0.25, 0.35, 0.5, 0.6):
    plt.plot(eta, resonance(eta, D), label=f"$D$={D:.2f}")

# Darstellung der Messwerte
plt.errorbar(etam, Am, yerr=Aerr, fmt='ko')

plt.title('Resonanzkurve')
plt.xlabel(r'$\eta=\omega/\omega_0$', size='x-large')
plt.ylabel('(normierte) Amplidude $A$')

plt.legend()
plt.show()
````
